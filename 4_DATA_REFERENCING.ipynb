<h4 style="color: #5d4e37; margin: 25px 0 15px 0; width: 100%; max-width: 100%; word-wrap: break-word; overflow-wrap: break-word; hyphens: auto; box-sizing: border-box; font-weight: 800;"><strong>DVL (Doppler Velocity Log) Rotation</strong></h4>

<p style="text-align: justify; margin: 15px 0; color: #4a4a4a; width: 100%; max-width: 100%; word-wrap: break-word; overflow-wrap: break-word; hyphens: auto; box-sizing: border-box;">
The DVL measures the robot's velocity. The table shows it is mounted with a 180° yaw, meaning it faces backward. When the robot moves forward, the DVL reports a backward velocity. To correct this, we rotate the DVL's measurements by 180°.
</p>

<div style="background-color: #f8f6f0; padding: 15px; border-left: 4px solid #8b7355; margin: 20px 0; width: 100%; max-width: 100%; box-sizing: border-box; overflow: hidden;">
<p style="margin: 0; color: #4a4a4a; word-wrap: break-word; overflow-wrap: break-word;">The equation for this transformation is:</p>
<div style="margin: 10px 0;">
  <math display="block">
    <mstyle mathcolor="#111827" mathvariant="bold">
      <mrow>
        <msub>
          <mi>V</mi>
          <mtext>body</mtext>
        </msub>
        <mo>=</mo>
        <msub>
          <mi>Q</mi>
          <mrow><mtext>DVL</mtext><mo>&#x2192;</mo><mi>B</mi></mrow>
        </msub>
        <mo>&#x2062;</mo>
        <msub>
          <mi>V</mi>
          <mtext>DVL</mtext>
        </msub>
        <mo>&#x2062;</mo>
        <msubsup>
          <mi>Q</mi>
          <mrow><mtext>DVL</mtext><mo>&#x2192;</mo><mi>B</mi></mrow>
          <mo>&#x2217;</mo>
        </msubsup>
      </mrow>
    </mstyle>
  </math>
</div>
<p style="margin: 10px 0 0 0; color: #4a4a4a; font-size: clamp(0.8em, 2vw, 0.9em); word-wrap: break-word; overflow-wrap: break-word;">Where <math display="inline"><msub><mi>Q</mi><mrow><mtext>DVL</mtext><mo>&#x2192;</mo><mi>B</mi></mrow></msub></math> is the quaternion for the 180° rotation from the DVL's frame to the robot's body frame.</p>
</div>

<h4 style="color: #5d4e37; margin: 25px 0 15px 0; width: 100%; max-width: 100%; word-wrap: break-word; overflow-wrap: break-word; hyphens: auto; box-sizing: border-box; font-weight: 800;"><strong>IMU (Inertial Measurement Unit) Rotation</strong></h4>

<p style="text-align: justify; margin: 15px 0; color: #4a4a4a; width: 100%; max-width: 100%; word-wrap: break-word; overflow-wrap: break-word; hyphens: auto; box-sizing: border-box;">
The IMU measures the robot's orientation. It is mounted with a 90° yaw, meaning it is turned sideways relative to the robot's front. We must correct for this 90° offset to get the robot's true orientation.
</p>

<p style="text-align: justify; margin: 15px 0; color: #4a4a4a; width: 100%; max-width: 100%; word-wrap: break-word; overflow-wrap: break-word; hyphens: auto; box-sizing: border-box;">
Additionally, IMUs often report data in a standard "aerial" coordinate system (East, North, Up). For underwater robotics, the "marine" frame (North, West, Down) is standard. Therefore, we apply two rotations to the IMU data:
</p>

<ol style="margin: 20px 0; color: #4a4a4a; width: 100%; max-width: 100%; word-wrap: break-word; overflow-wrap: break-word; box-sizing: border-box; padding-left: 20px;">
  <li style="margin: 8px 0; word-wrap: break-word; overflow-wrap: break-word;"><strong>Frame Conversion:</strong> We use <code style="background-color: #f0efe8; padding: 2px 4px; border-radius: 3px; color: #5d4e37; word-wrap: break-word;">Q_TER_MARIN</code> to convert the data from the aerial frame to the marine frame.</li>
  <li style="margin: 8px 0; word-wrap: break-word; overflow-wrap: break-word;"><strong>Mounting Correction:</strong> We then use <code style="background-color: #f0efe8; padding: 2px 4px; border-radius: 3px; color: #5d4e37; word-wrap: break-word;">Q_IMU_B</code> to correct for the 90° mounting offset, aligning the sensor's orientation with the robot's body frame.</li>
</ol>

<p style="text-align: justify; margin: 20px 0; color: #4a4a4a; font-style: italic; width: 100%; max-width: 100%; word-wrap: break-word; overflow-wrap: break-word; hyphens: auto; box-sizing: border-box;">
These rotations ensure all sensor data accurately reflects the robot's movement and orientation in a single, consistent coordinate system.
</p>

</div>

```python
import numpy as np
from quat_utils import quatconj, quatmultiply 
from types import SimpleNamespace
from math import cos, sin, pi
from h5_utils import save, load
```

```python
DVL_VALID      = load('3_VALID_DATA.h5', 'DVL_VALID')
ECHO_VALID     = load('3_VALID_DATA.h5', 'ECHO_VALID')
IMU_VALID_QUAT = load('3_VALID_DATA.h5', 'IMU_VALID_QUAT')
PING360_VALID  = load('3_VALID_DATA.h5', 'PROFILO_PING360_VALID')
SEAKING_VALID  = load('3_VALID_DATA.h5', 'PROFILO_SEAKING_VALID')
```

<div style="font-family: 'Times New Roman', Georgia, serif; background-color: #fefdf8; padding: 20px; border: 2px solid #d4c4a8; border-radius: 8px; box-shadow: 0 4px 8px rgba(0,0,0,0.1); line-height: 1.6; max-width: 100%; box-sizing: border-box; overflow: hidden;">

<h3 style="color: #5d4e37; text-align: center; margin-bottom: 20px; font-weight: bold; max-width: 100%; word-wrap: break-word; hyphens: auto;">Define Rotation Quaternions</h3>

<p style="text-align: justify; margin: 20px 0; color: #4a4a4a; width: 100%; max-width: 100%; word-wrap: break-word; overflow-wrap: break-word; hyphens: auto; box-sizing: border-box;">
This cell defines the constant quaternions that will be used to perform the rotations. These quaternions represent the fixed orientation offsets of the DVL and IMU sensors relative to the robot's body frame, as well as the transformation from the terrestrial to the marine reference frame.
</p>

</div>

```python
Q_DVL_B     = [cos(pi/2), 0, 0, sin(pi/2)] # 180° rotation about the Z-axis.
Q_TER_MARIN = [cos(pi/2), sin(pi/2), 0, 0] # 180° rotation about the X-axis.
Q_IMU_B     = [cos(pi/4), 0, 0, sin(pi/4)] # 90° rotation about the Z-axis.
```

<div style="font-family: 'Times New Roman', Georgia, serif; background-color: #fefdf8; padding: 20px; border: 2px solid #d4c4a8; border-radius: 8px; box-shadow: 0 4px 8px rgba(0,0,0,0.1); line-height: 1.6; max-width: 100%; box-sizing: border-box; overflow: hidden;">

<h3 style="color: #5d4e37; text-align: center; margin-bottom: 20px; font-weight: bold; max-width: 100%; word-wrap: break-word; hyphens: auto;">Initialize Referenced Data Structure</h3>

<p style="text-align: justify; margin: 20px 0; color: #4a4a4a; width: 100%; max-width: 100%; word-wrap: break-word; overflow-wrap: break-word; hyphens: auto; box-sizing: border-box;">
A new data structure, <code style="background-color: #f0efe8; padding: 2px 4px; border-radius: 3px; color: #5d4e37; word-wrap: break-word;">REF_DATA</code>, is initialized to store the results of the frame transformations. It contains arrays to hold the referenced velocity, depth, and orientation data.
</p>

</div>

```python
n = len(DVL_VALID.ABS_TIME)
m = len(ECHO_VALID.ABS_TIME)
k = len(IMU_VALID_QUAT.ABS_TIME) ## Number of echo samples

REF_DATA = SimpleNamespace(
    VEL_B                       = np.zeros((3, n)),
    VEL_B_TIME_STAMP            = np.zeros(n),
    DEPTH_0                     = np.zeros(n),
    DEPTH_0_TIME_STAMP          = np.zeros(n),
    QUATERNION_ECHO             = np.zeros((m,4)),
    QUATERNION_ECHO_TIME_STAMP  = np.zeros(m),
    QUATERNION_IMU              = np.zeros((k,4)),
    QUATERNION_IMU_TIME_STAMP   = np.zeros(k),
)
```

<div style="font-family: 'Times New Roman', Georgia, serif; background-color: #fefdf8; padding: 20px; border: 2px solid #d4c4a8; border-radius: 8px; box-shadow: 0 4px 8px rgba(0,0,0,0.1); line-height: 1.6; max-width: 100%; box-sizing: border-box; overflow: hidden;">

<h3 style="color: #5d4e37; text-align: center; margin-bottom: 20px; font-weight: bold; max-width: 100%; word-wrap: break-word; hyphens: auto;">Perform Frame Transformations</h3>

<p style="text-align: justify; margin: 20px 0; color: #4a4a4a; width: 100%; max-width: 100%; word-wrap: break-word; overflow-wrap: break-word; hyphens: auto; box-sizing: border-box;">
This is the core processing cell where the sensor data is transformed into the common body frame.
</p>

<ul style="margin: 20px 0; color: #4a4a4a; width: 100%; max-width: 100%; word-wrap: break-word; overflow-wrap: break-word; box-sizing: border-box; padding-left: 20px;">
  <li style="margin: 8px 0; word-wrap: break-word; overflow-wrap: break-word;"><strong>DVL:</strong> The velocity data is rotated to account for the sensor's backward-facing orientation.</li>
  <li style="margin: 8px 0; word-wrap: break-word; overflow-wrap: break-word;"><strong>Echosounder:</strong> The orientation quaternions are normalized.</li>
  <li style="margin: 8px 0; word-wrap: break-word; overflow-wrap: break-word;"><strong>IMU:</strong> The orientation quaternions are first converted from the aerial (ENU) to the marine (NWD) frame and then rotated to correct for the sensor's mounting offset.</li>
</ul>

</div>

```python
for i in range(n):
    vq = [0, DVL_VALID.U[i], DVL_VALID.V[i], DVL_VALID.W[i]]             # velocity quaternion
    Q_TEMP = quatmultiply(Q_DVL_B, quatmultiply(vq, quatconj(Q_DVL_B)))  # Expressing the velocity in the robot's frame
    REF_DATA.VEL_B[:, i]           = Q_TEMP[1:4]                         # the velocty after being rotated 180 degrees around z
    REF_DATA.VEL_B_TIME_STAMP[i]   = DVL_VALID.ABS_TIME[i]
    REF_DATA.DEPTH_0[i]            = -DVL_VALID.DEPTH[i]                 # For the graph with the  trajectories, when we see the figure we want the trajectory directed down when the robot is moving downwards otherwise it will seems like going up
    REF_DATA.DEPTH_0_TIME_STAMP[i] = DVL_VALID.ABS_TIME[i]


for i in range(m):
    REF_DATA.QUATERNION_ECHO[i, :]         = ECHO_VALID.QUAT[i,:] / np.linalg.norm(ECHO_VALID.QUAT[i,:]) # Normalization
    REF_DATA.QUATERNION_ECHO_TIME_STAMP[i] = ECHO_VALID.ABS_TIME[i]


for i in range(k):
    normalized_quat = IMU_VALID_QUAT.QUAT[i,:] / np.linalg.norm(IMU_VALID_QUAT.QUAT[i,:]) # Normalization
    
    CURRENT_QUAT                          = (normalized_quat[0], normalized_quat[1], normalized_quat[2], normalized_quat[3])
    CURRENT_QUAT                          = quatmultiply(Q_TER_MARIN, CURRENT_QUAT)                                           # expressing IMU's quaternons from aerial frame to marine frame 
    REF_DATA.QUATERNION_IMU[i, :]         = quatmultiply(CURRENT_QUAT, Q_IMU_B)                      # IMU's quaternions now are expressed like robot's frame (W)est N(oth) D(own) == NWD ///  expressing IMU's quaternions / orientation by marine / global  frame to robot's frame
    REF_DATA.QUATERNION_IMU_TIME_STAMP[i] = IMU_VALID_QUAT.ABS_TIME[i]
```

```python
save('HDF5_DATA/4_REFERENCED_DATA.h5', 'REF_DATA', REF_DATA, mode='w')
```
