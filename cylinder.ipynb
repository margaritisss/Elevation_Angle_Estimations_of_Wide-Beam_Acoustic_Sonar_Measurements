```python
import numpy as np
import matplotlib.pyplot as plt

# %matplotlib notebook # Uncomment if you want an interactive plot in classic Jupyter

fig = plt.figure(figsize=(12, 9))
ax = fig.add_subplot(111, projection='3d')

# 1. Define the parameters for the elliptic cylinder
a = 5.0       # Semi-major axis (x-direction)
b = 2.0       # Semi-minor axis (y-direction)
height = 8.0  # Total height (Length) of the cylinder

# 2. Create the data grid for the cylinder surface
z = np.linspace(-height/2, height/2, 50)
theta = np.linspace(0, 2*np.pi, 50)
theta_grid, z_grid = np.meshgrid(theta, z)

# Parametric equations for an ellipse
x_grid = a * np.cos(theta_grid)
y_grid = b * np.sin(theta_grid)

# 3. Plot the elliptic cylinder surface
ax.plot_surface(x_grid, y_grid, z_grid, alpha=0.2, color='cyan', edgecolor='none')

# 4. Plot the core geometric axes
ax.plot([0, 0], [0, 0], [-height/2, height/2], color='red', linewidth=2, linestyle='--', label='Cylinder Axis')

top_z = height/2
# Major Axis (Max Width)
ax.plot([-a, a], [0, 0], [top_z, top_z], color='blue', linewidth=3, label='Major Axis (Max Width = 2a)')
# Minor Axis (Min Width)
ax.plot([0, 0], [-b, b], [top_z, top_z], color='green', linewidth=3, label='Minor Axis (Min Width = 2b)')


# --- NEW ADDITIONS FOR LENGTH AND WIDTH VISUALIZATION ---

# Plot a line indicating the overall Length (Height) on the outside of the cylinder
ax.plot([a, a], [0, 0], [-height/2, height/2], color='purple', linewidth=4, label='Length of Cylinder')

# Add Text Annotations directly on the 3D plot to make it extremely clear
# Annotate Length
ax.text(a + 0.5, 0, 0, "LENGTH\n(Height)", color='purple', fontsize=10, fontweight='bold')

# Annotate Max Width (Major Axis)
ax.text(0, 0, top_z + 0.5, "MAX WIDTH\n(Major Axis)", color='blue', fontsize=10, fontweight='bold', ha='center')

# Annotate Min Width (Minor Axis)
ax.text(0, b + 0.5, top_z, "MIN WIDTH\n(Minor Axis)", color='green', fontsize=10, fontweight='bold')

# --------------------------------------------------------

# Formatting the plot
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')
ax.set_title('Elliptic Cylinder: Defining Length and Width')
ax.legend(loc='upper left')

# Set axis limits to be roughly equal to prevent visual distortion
max_limit = max(a, height/2) + 1
ax.set_xlim([-max_limit, max_limit])
ax.set_ylim([-max_limit, max_limit])
ax.set_zlim([-max_limit, max_limit])

plt.show()
```
